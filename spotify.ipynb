```python
import zipfile
import os

# Extract the ZIP file
with zipfile.ZipFile('/content/archive.zip', 'r') as zip_ref:
    zip_ref.extractall('/content/spotify_data')

print("Dataset extracted successfully!")

# Display extracted files
print(os.listdir('/content/spotify_data'))
```

```python
import pandas as pd
import glob
import os

# Find the CSV file inside the extracted folder
csv_files = glob.glob('/content/spotify_data/**/*.csv', recursive=True)

# Read the CSV file
df = pd.read_csv(csv_files[0])

# Display first 5 rows
print("FIRST 5 ROWS OF THE DATASET")
display(df.head())

# Display dataset dimensions
print("\nDATASET SHAPE:")
print(df.shape)

# Display column names
print("\nCOLUMN NAMES:")
print(df.columns.tolist())
```

```python
# Display all column names
print("ALL COLUMN NAMES:")
print(df.columns.tolist())


# Understand data types and non-null values
print("\nDATASET INFORMATION:")
df.info()


# Check missing values in each column
print("\nMISSING VALUES:")
print(df.isnull().sum())


# Statistical summary of numerical columns
print("\nSTATISTICAL SUMMARY:")
display(df.describe())
```

```python
# DATA TRANSFORMATION

# Remove unnecessary unnamed index column
df.drop(columns=['Unnamed: 0'], inplace=True, errors='ignore')

# Convert duration from milliseconds to minutes
df['duration_min'] = df['duration_ms'] / 60000

# Create a new popularity category
df['popularity_category'] = df['popularity'].apply(
    lambda x: 'Popular' if x >= 70 else 'Less Popular'
)

# Display the transformed dataset
print("DATA AFTER TRANSFORMATION:")
display(df.head())

print("\nUPDATED DATASET SHAPE:")
print(df.shape)
```

```python
# CHECK MISSING VALUES BEFORE HANDLING

print("Missing values before handling:")
print(df.isnull().sum())

# Select numerical columns
numeric_columns = df.select_dtypes(
    include='number'
).columns

# Apply mean imputation
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].mean())

# Handle missing text values
text_columns = df.select_dtypes(
    include='object'
).columns

for column in text_columns:
    df[column] = df[column].fillna('Unknown')

# Check missing values after handling
print("\nMissing values after handling:")
print(df.isnull().sum())

print("\nNULL VALUE HANDLING COMPLETED!")
```

```python
# INSIGHT GENERATION

# 1. Average song popularity
average_popularity = df['popularity'].mean()

print("1. Average Song Popularity:", round(average_popularity, 2))


# 2. Top 10 most popular songs
top_songs = df.nlargest(10, 'popularity')

print("\n2. TOP 10 MOST POPULAR SONGS:")
display(top_songs[['track_name', 'artists', 'popularity']])


# 3. Top 10 artists by average popularity
top_artists = (
    df.groupby('artists')['popularity']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

print("\n3. TOP 10 ARTISTS BY AVERAGE POPULARITY:")
display(top_artists)


# 4. Top 10 genres by average popularity
top_genres = (
    df.groupby('track_genre')['popularity']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

print("\n4. TOP 10 GENRES BY AVERAGE POPULARITY:")
display(top_genres)


# 5. Average song duration
average_duration = df['duration_min'].mean()

print("\n5. Average Song Duration:",
      round(average_duration, 2), "minutes")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")


# 1. Top 10 Popular Songs
plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_songs,
    x='popularity',
    y='track_name'
)

plt.title("Top 10 Most Popular Songs")
plt.xlabel("Popularity")
plt.ylabel("Song Name")
plt.tight_layout()
plt.show()


# 2. Top 10 Popular Genres
plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_genres.values,
    y=top_genres.index
)

plt.title("Top 10 Genres by Average Popularity")
plt.xlabel("Average Popularity")
plt.ylabel("Music Genre")
plt.tight_layout()
plt.show()


# 3. Distribution of Song Popularity
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x='popularity',
    bins=20
)

plt.title("Distribution of Song Popularity")
plt.xlabel("Popularity")
plt.ylabel("Number of Songs")
plt.tight_layout()
plt.show()


# 4. Danceability vs Popularity
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df.sample(5000, random_state=42),
    x='danceability',
    y='popularity',
    alpha=0.5
)

plt.title("Danceability vs Song Popularity")
plt.xlabel("Danceability")
plt.ylabel("Popularity")
plt.tight_layout()
plt.show()
```

```python
# Calculate correlation between danceability and popularity

correlation = df['danceability'].corr(df['popularity'])

print("Correlation between Danceability and Popularity:",
      round(correlation, 3))

if correlation > 0.5:
    print("There is a strong positive relationship.")
elif correlation > 0:
    print("There is a weak positive relationship.")
elif correlation < -0.5:
    print("There is a strong negative relationship.")
else:
    print("There is a weak or negligible relationship.")
```
